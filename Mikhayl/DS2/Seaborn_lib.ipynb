```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
data = [10,20,25,30,40,45,50]
```

```python
sns.lineplot(data)
plt.show()
```

```python
students = ['Aman','Riya','Aoun','Mikhayl']
marks = [80,90,75,95]

sns.barplot(x=students, y=marks)
plt.show()
```

```python
hours = [1,2,3,4,5,6]
marks = [40,45,50,60,70,80]

sns.scatterplot(x=hours, y=marks)

plt.show()
```

```python
marks = [45,50,52,55,60,60,65,70,72,75,80,85,90]

sns.histplot(marks, bins=10)

plt.show()
```

```python
students = ['male','female','male','male','female','female','male']

sns.countplot(x=students)
plt.show()
```
