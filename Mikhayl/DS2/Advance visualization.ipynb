# Advance Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 65, 68, 70, 72, 75, 75, 78, 80, 82, 85, 90, 92, 95]


# marks.mean()
total = 0

for value in marks:
    total += value
# print(total)
    
average = total / len(marks)

print(average)
```

## Box plots

```python
plt.boxplot(marks)
plt.ylabel("Marks")
plt.title("Student Marks Box Plot")
plt.show()
```

## histograms

```python
plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Student Marks")
plt.show()
```

## heatmap

```python
import seaborn as sns

data = {
    "Math":[80,70,90,60,85],
    "Science":[75,65,95,55,80],
    "English":[70, 80, 85, 75, 90],
    "Computer":[90, 75, 95, 60, 85]
}

df = pd.DataFrame(data)
correlation = df.corr()
sns.heatmap(correlation, annot=True)
plt.show()
```
