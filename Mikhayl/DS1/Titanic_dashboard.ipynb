```python
import pandas as pd
import plotly.express as px
import seaborn as sns

df = sns.load_dataset('titanic')
df.head()
```

# Survival count

```python
fig = px.bar(df, x='sex', color='survived',title='Survival count by gender',labels={'survived':'Survived (0-No, 1=Yes)'})
fig.show()
```

# Age distribution

```python
fig = px.histogram(df, x='age', nbins=30, title='Age distribution of Passengers')
fig.show()
```
