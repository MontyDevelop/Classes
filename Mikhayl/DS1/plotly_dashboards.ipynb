```python
import plotly.express as px
```

```python
x = [1,2,3,4,5]
y = [10,15,18,8,20]
```

```python
fig = px.line(x=x, y=y, title='Simple line Chart')
fig.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
fig  = make_subplots(rows=1, cols=2, subplot_titles=("Line chart", "Bar Chart"))

fig.add_trace(go.Scatter(x=[1,2,3,4,5], y=[10,20,7,15]), row=1, col=1)
fig.add_trace(go.Bar(x=['A','B','C','D'], y=[5,7,3,8]), row=1, col=2)
fig.update_layout(title='Simple Plotly Dashboard')
fig.show()
```

```python
import plotly.express as px
import pandas as pd

data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "Sales": [200, 350, 300, 400, 380]
}

df = pd.DataFrame(data)

fig = px.bar(df, x="Day", y="Sales", title="Weekly Sales Dashboard")
fig.show()
```
